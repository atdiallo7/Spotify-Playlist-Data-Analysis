# Identifying Songs with missing Genres

### Import Dependencies


```python
import pandas as pd
import numpy as np
```

### Loading Data 

```python
url = "https://docs.google.com/spreadsheets/d/19ateaoZLzZdsXk35COqZ0yT9bKq01gTe7EtTM5_unFE/export?format=csv"
df = pd.read_csv(url)
```

### Creating new DataFrame for songs with missing Genres

```python
# exclude useless columns
relevant_df = df.loc[:, ~df.columns.isin(["Track URI", "Added By", "Added At"])]
relevant_df.head()
df_nan = relevant_df[pd.isna(relevant_df["Genres"])]
df_nan.head()
```
