# Song Clustering Analysis

Goal: Group songs into clusters that match or refine the existing genre labels.


```python
import pandas as pd

amadou = pd.read_csv("../data/Amadou Music Taste Export.csv")
logan = pd.read_csv("../data/Logan Exported Data.csv")
```

```python
amadou["Listener"] = "Amadou"
logan["Listener"] = "Logan"

data = pd.concat([amadou, logan], ignore_index=True)
```

```python
print("kernel working")
```

```python
print(data.shape)
print(data["Listener"].value_counts())
```

```python
features = [
    "Danceability",
    "Energy",
    "Valence",
    "Acousticness",
    "Tempo",
    "Loudness"
]

data[features].isna().sum()
```

```python
cluster_data = data.dropna(subset=features).copy()

print(cluster_data.shape)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
from sklearn.preprocessing import StandardScaler

X = cluster_data[features]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print(X_scaled.shape)
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

scores = {}

for k in range(2, 9):
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = model.fit_predict(X_scaled)
    scores[k] = silhouette_score(X_scaled, labels)

scores
```
